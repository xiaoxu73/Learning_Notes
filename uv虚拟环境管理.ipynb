````python
# 查找所有python
!uv python list
!uv python find 3.10  # 查找所有3.10版本的Python解释器
````

````python
# 初始化项目（默认用系统Python版本）
!uv init # 会自动创建.venv虚拟环境目录
# 初始化指定Python版本的项目
!uv init --python 3.10
````

### 当前项目目录结构

```
my_uv_project/
├── .gitignore        # Git忽略文件（自动包含.venv等环境目录）
├── .python-version   # 记录项目Python版本（类似pyenv）
├── main.py           # 示例入口脚本（可以删掉自己写）
├── pyproject.toml    # 项目配置文件（核心！记录依赖和设置）
└── README.md         # 项目说明文档
```

````python
# 激活当前项目的虚拟环境（也可以直接用uv run执行脚本，无需激活）
!uv activate
# 不用激活环境，直接用uv run执行脚本，UV也会自动使用项目的虚拟环境
!uv run main.py  # 自动用.venv里的Python执行main.py
````

````python
# 根据pyproject.toml同步安装依赖（安装新增的，删除多余的）
!uv sync
# 更新锁文件（不修改pyproject.toml，只更新uv.lock）
!uv lock
````

````python
!uv tree  # 显示所有依赖的树形结构
!uv tree pandas  # 只显示pandas及其依赖
````

````python
# 添加单个依赖（默认最新版）
!uv add pandas
# 添加指定版本
!uv add requests==2.31.0
# 添加开发环境依赖（仅开发时用，比如pytest）
!uv add pytest --dev  # 会记录在pyproject.toml的[project.optional-dependencies.dev]里
# 更新单个包到最新版
!uv add pandas --upgrade  # 和uv update pandas效果一样
# 更新所有依赖到最新版
!uv remove pandas  # 从依赖中删除pandas，同时更新pyproject.toml和uv.lock
````
